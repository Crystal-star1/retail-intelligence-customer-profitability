# E-Commerce Retail Intelligence & Customer RFM Segmentation Analysis

## Project Overview
This project provides a comprehensive retail intelligence analysis of an e-commerce platform's operations. By consolidating transaction, customer, product, and category datasets, we construct a high-performance analytical pipeline using DuckDB and pandas. The ultimate objective is to engineer a unified transaction-level data mart (`fact_sales`), generate sub-aggregations across key dimensions, and implement an RFM (Recency, Frequency, Monetary) Customer Segmentation engine to drive targeted marketing and business growth strategies.

## Dataset Overview
The analysis is powered by seven relational datasets in Parquet format:
1. **`customers.parquet`**: Customer profile details including signup dates, location, loyalty tiers, and marketing preferences.
2. **`orders.parquet`**: Order-level metadata including order date, status, channel, and shipping information.
3. **`order_items.parquet`**: Transaction details mapping products to orders with quantity and pricing metrics.
4. **`products.parquet`**: Catalog of items, cost prices, SKUs, and lifecycle status.
5. **`categories.parquet`**: Hierarchy of product categories and parent relations.
6. **`payments.parquet`**: Transaction payment methods, amounts, and statuses.
7. **`returns.parquet`**: Post-purchase product return logs, reasons, and refund amounts.

**Reasoning**:
We need to satisfy instruction 2, 3, and 4 of the subtask by consolidating imports, handling suffix-copied parquet files, verifying file existence, and creating a clean DuckDB connection in a single code block.



```python
import os
import shutil
from pathlib import Path
import pandas as pd
import duckdb

# 1. Clean up file names with suffix copies if they exist
expected_files = [
    "categories.parquet",
    "customers.parquet",
    "products.parquet",
    "orders.parquet",
    "order_items.parquet",
    "payments.parquet",
    "returns.parquet"
]

# Look for files matching pattern like 'filename (number).parquet' and copy them to 'filename.parquet'
for file_name in expected_files:
    if not os.path.exists(file_name):
        stem = Path(file_name).stem
        suffix = Path(file_name).suffix
        # Search current working directory for matching files with suffix
        for f in os.listdir("."):
            if f.startswith(stem) and f.endswith(suffix) and f != file_name:
                shutil.copyfile(f, file_name)
                print(f"Copied {f} to {file_name}")
                break

# 2. Verify all expected files exist
missing_files = [f for f in expected_files if not os.path.exists(f)]
if missing_files:
    print(f"Warning: The following files are still missing: {missing_files}")
else:
    print("All expected Parquet files are present in the workspace.")

# 3. Initialize DuckDB Database Connection
con = duckdb.connect("retail_intelligence.duckdb")
print("Successfully connected to retail_intelligence.duckdb!")
```

### Building the Consolidated Sales View
This cell creates the `fact_sales` view, which integrates orders, products, and categories, while calculating financial metrics like gross profit and net revenue.

```python
con.execute("""
CREATE OR REPLACE VIEW fact_sales AS
SELECT
    oi.order_item_id,
    o.order_id,
    o.ordered_at,
    o.shipped_at,
    o.status AS order_status,
    o.channel,
    o.shipping_country,
    o.coupon_code,
    c.customer_id,
    c.city,
    c.country,
    c.signup_date,
    c.loyalty_tier,
    c.marketing_opt_in,
    p.product_id,
    p.sku,
    p.product_name,
    p.discontinued_at,
    cat.category_id,
    cat.category_name,
    cat.parent_category_id,
    oi.quantity,
    CAST(oi.unit_price AS DECIMAL(12,2)) AS sale_price,
    CAST(oi.discount_amount AS DECIMAL(12,2)) AS discount_amount,
    CAST(p.unit_cost AS DECIMAL(12,2)) AS unit_cost,
    (oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) AS gross_revenue,
    ((oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) - CAST(oi.discount_amount AS DECIMAL(12,2))) AS net_revenue,
    (oi.quantity * CAST(p.unit_cost AS DECIMAL(12,2))) AS cost,
    (((oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) - CAST(oi.discount_amount AS DECIMAL(12,2))) - (oi.quantity * CAST(p.unit_cost AS DECIMAL(12,2)))) AS gross_profit
FROM order_items oi
JOIN orders o ON oi.order_id = o.order_id
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON oi.product_id = p.product_id
JOIN categories cat ON p.category_id = cat.category_id
""")
print("Consolidated fact_sales view created successfully.")
```

### Generating Analytical Sub-aggregations
This cell builds summarized analytical views focusing on performance broken down by customers, individual products, and product categories.

```python
con.execute("""
CREATE OR REPLACE VIEW customer_metrics AS
SELECT
    customer_id,
    country,
    loyalty_tier,
    COUNT(DISTINCT order_id) AS orders,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    SUM(quantity) AS items,
    MIN(ordered_at) AS first_order,
    MAX(ordered_at) AS last_order,
    SUM(net_revenue) / COUNT(DISTINCT order_id) AS aov
FROM fact_sales
GROUP BY customer_id, country, loyalty_tier;
""")

con.execute("""
CREATE OR REPLACE VIEW product_metrics AS
SELECT
    product_id,
    product_name,
    category_name,
    COUNT(DISTINCT order_id) AS orders,
    SUM(quantity) AS units_sold,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    AVG(net_revenue) AS avg_line_revenue
FROM fact_sales
GROUP BY product_id, product_name, category_name;
""")

con.execute("""
CREATE OR REPLACE VIEW category_metrics AS
SELECT
    category_id,
    category_name,
    COUNT(DISTINCT product_id) AS products,
    COUNT(DISTINCT order_id) AS orders,
    SUM(quantity) AS units_sold,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    SUM(gross_profit) / NULLIF(SUM(net_revenue), 0) AS profit_margin
FROM fact_sales
GROUP BY category_id, category_name;
""")

print("Metrics views for Customers, Products, and Categories initialized.")
```

### RFM Segmentation Engine
This cell reads customer metrics into pandas, computes Recency, Frequency, and Monetary scores using quantiles, and classifies customers into cohorts.

```python
import pandas as pd

rfm = con.sql("""
SELECT
    customer_id,
    MAX(ordered_at) AS last_order_date,
    COUNT(DISTINCT order_id) AS frequency,
    SUM(net_revenue) AS monetary
FROM fact_sales
GROUP BY customer_id
""").df()

snapshot_date = rfm["last_order_date"].max()
rfm["recency"] = (snapshot_date - rfm["last_order_date"]).dt.days

rfm["R"] = pd.qcut(rfm["recency"], 5, labels=[5, 4, 3, 2, 1])
rfm["F"] = pd.qcut(rfm["frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5])
rfm["M"] = pd.qcut(rfm["monetary"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5])

def assign_segment(row):
    r, f, m = int(row["R"]), int(row["F"]), int(row["M"])
    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"
    elif r >= 3 and f >= 3 and m >= 3:
        return "Loyal Customers"
    elif r >= 4 and f <= 2:
        return "Potential Loyalists"
    elif r <= 2 and f >= 3:
        return "At Risk"
    return "Others"

rfm["segment"] = rfm.apply(assign_segment, axis=1)
display(rfm["segment"].value_counts().to_frame())
```

```python
import pandas as pd

rfm = con.sql("""
SELECT
    customer_id,
    MAX(ordered_at) AS last_order_date,
    COUNT(DISTINCT order_id) AS frequency,
    SUM(net_revenue) AS monetary
FROM fact_sales
GROUP BY customer_id
""").df()

snapshot_date = rfm["last_order_date"].max()

rfm["recency"] = (
    snapshot_date - rfm["last_order_date"]
).dt.days

rfm["R"] = pd.qcut(
    rfm["recency"],
    5,
    labels=[5,4,3,2,1]
)

rfm["F"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    5,
    labels=[1,2,3,4,5]
)

rfm["M"] = pd.qcut(
    rfm["monetary"].rank(method="first"),
    5,
    labels=[1,2,3,4,5]
)

rfm["RFM_Score"] = (
    rfm["R"].astype(str)
    + rfm["F"].astype(str)
    + rfm["M"].astype(str)
)

def segment(row):

    r = int(row["R"])
    f = int(row["F"])
    m = int(row["M"])

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    if r >= 3 and f >= 3 and m >= 3:
        return "Loyal Customers"

    if r >= 4 and f <= 2:
        return "Potential Loyalists"

    if r <= 2 and f >= 3:
        return "At Risk"

    return "Others"

rfm["segment"] = rfm.apply(segment, axis=1)

rfm["segment"].value_counts()
```

```python
at_risk_revenue = (
    rfm.loc[
        rfm["segment"] == "At Risk",
        "monetary"
    ].sum()
)

total_revenue = rfm["monetary"].sum()

at_risk_pct = (
    at_risk_revenue / total_revenue
) * 100

print(f"At Risk Revenue: ${at_risk_revenue:,.2f}")
print(f"Revenue Share: {at_risk_pct:.2f}%")
print(f"At Risk Customers: {(rfm['segment'] == 'At Risk').sum():,}")
```

```python
champion_revenue = (
    rfm.loc[
        rfm["segment"] == "Champions",
        "monetary"
    ].sum()
)

total_revenue = rfm["monetary"].sum()

champion_pct = (
    champion_revenue / total_revenue
) * 100

print(f"Champion Revenue: ${champion_revenue:,.2f}")
print(f"Revenue Share: {champion_pct:.2f}%")
print(f"Champion Customers: {(rfm['segment'] == 'Champions').sum():,}")
```

## Phase 3: SQL Analytical Layer View Setup

### Subtask:
Initialize the analytical layer by defining structured SQL views in DuckDB.


**Reasoning**:
Create the analytical layer SQL views (fact_sales, customer_metrics, product_metrics, category_metrics, channel_metrics) in DuckDB with complete PII redaction (omitting first_name, last_name, and email).



```python
import duckdb

# Ensure connection is established
con = duckdb.connect('retail_intelligence.duckdb')

# 1. Create Consolidated fact_sales view with PII redacted (first_name, last_name, email omitted or replaced)
con.execute("""
CREATE OR REPLACE VIEW fact_sales AS
SELECT
    oi.order_item_id,
    o.order_id,
    o.ordered_at,
    o.shipped_at,
    o.status AS order_status,
    o.channel,
    o.shipping_country,
    o.coupon_code,
    c.customer_id,
    c.city,
    c.country,
    c.signup_date,
    c.loyalty_tier,
    c.marketing_opt_in,
    p.product_id,
    p.sku,
    p.product_name,
    p.discontinued_at,
    cat.category_id,
    cat.category_name,
    cat.parent_category_id,
    oi.quantity,
    CAST(oi.unit_price AS DECIMAL(12,2)) AS sale_price,
    CAST(oi.discount_amount AS DECIMAL(12,2)) AS discount_amount,
    CAST(p.unit_cost AS DECIMAL(12,2)) AS unit_cost,
    (oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) AS gross_revenue,
    ((oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) - CAST(oi.discount_amount AS DECIMAL(12,2))) AS net_revenue,
    (oi.quantity * CAST(p.unit_cost AS DECIMAL(12,2))) AS cost,
    (((oi.quantity * CAST(oi.unit_price AS DECIMAL(12,2))) - CAST(oi.discount_amount AS DECIMAL(12,2))) - (oi.quantity * CAST(p.unit_cost AS DECIMAL(12,2)))) AS gross_profit
FROM order_items oi
JOIN orders o ON oi.order_id = o.order_id
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON oi.product_id = p.product_id
JOIN categories cat ON p.category_id = cat.category_id
""")

# 2. Create customer_metrics view grouping by non-PII details
con.execute("""
CREATE OR REPLACE VIEW customer_metrics AS
SELECT
    customer_id,
    country,
    loyalty_tier,
    COUNT(DISTINCT order_id) AS orders,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    SUM(quantity) AS items,
    MIN(ordered_at) AS first_order,
    MAX(ordered_at) AS last_order,
    SUM(net_revenue) / NULLIF(COUNT(DISTINCT order_id), 0) AS aov
FROM fact_sales
GROUP BY customer_id, country, loyalty_tier
""")

# 3. Create product_metrics view
con.execute("""
CREATE OR REPLACE VIEW product_metrics AS
SELECT
    product_id,
    product_name,
    category_name,
    COUNT(DISTINCT order_id) AS orders,
    SUM(quantity) AS units_sold,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    AVG(net_revenue) AS avg_line_revenue
FROM fact_sales
GROUP BY product_id, product_name, category_name
""")

# 4. Create category_metrics view
con.execute("""
CREATE OR REPLACE VIEW category_metrics AS
SELECT
    category_id,
    category_name,
    COUNT(DISTINCT product_id) AS products,
    COUNT(DISTINCT order_id) AS orders,
    SUM(quantity) AS units_sold,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    SUM(gross_profit) / NULLIF(SUM(net_revenue), 0) AS profit_margin
FROM fact_sales
GROUP BY category_id, category_name
""")

# 5. Create channel_metrics view
con.execute("""
CREATE OR REPLACE VIEW channel_metrics AS
SELECT
    channel,
    COUNT(DISTINCT order_id) AS orders,
    COUNT(DISTINCT customer_id) AS customers,
    SUM(quantity) AS units_sold,
    SUM(net_revenue) AS revenue,
    SUM(gross_profit) AS profit,
    SUM(net_revenue) / NULLIF(COUNT(DISTINCT order_id), 0) AS aov,
    SUM(gross_profit) / NULLIF(SUM(net_revenue), 0) AS profit_margin
FROM fact_sales
GROUP BY channel
""")

print("All requested SQL analytical views successfully created and verified in DuckDB!")
```

## Phase 5: Customer Segmentation & RFM Analysis

### Subtask:
Build the quantile-based RFM scoring pipeline, analyze customer segment distributions, calculate revenue concentration metrics, and export clean Excel assets.


**Reasoning**:
We will run the RFM Customer Segmentation analysis in pandas using the views from DuckDB, compute the R, F, and M quintiles, classify customers into cohorts, evaluate the revenue concentration, and save the required outputs to clean Excel sheets.



```python
import duckdb
import pandas as pd
import numpy as np

# 1. Connect to the existing DuckDB database
con = duckdb.connect('retail_intelligence.duckdb')

# 2. Extract metrics from customer_metrics and fact_sales
rfm_raw = con.sql("""
SELECT
    customer_id,
    MAX(ordered_at) AS last_order_date,
    COUNT(DISTINCT order_id) AS frequency,
    SUM(net_revenue) AS monetary
FROM fact_sales
GROUP BY customer_id
""").df()

# Ensure last_order_date is datetime
rfm_raw['last_order_date'] = pd.to_datetime(rfm_raw['last_order_date'])
snapshot_date = rfm_raw['last_order_date'].max()
rfm_raw['recency'] = (snapshot_date - rfm_raw['last_order_date']).dt.days

# 3. Calculate quintile scores
# R: lower recency is better (score 5)
rfm_raw['R'] = pd.qcut(rfm_raw['recency'], 5, labels=[5, 4, 3, 2, 1])
# F and M: higher frequency/monetary is better
rfm_raw['F'] = pd.qcut(rfm_raw['frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])
rfm_raw['M'] = pd.qcut(rfm_raw['monetary'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])

# 4. Map customers to cohorts
def assign_cohort(row):
    r, f, m = int(row['R']), int(row['F']), int(row['M'])
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif r >= 3 and f >= 3 and m >= 3:
        return 'Loyal Customers'
    elif r >= 4 and f <= 2:
        return 'Potential Loyalists'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    return 'Others'

rfm_raw['segment'] = rfm_raw.apply(assign_cohort, axis=1)

# 5. Calculate and display segment distribution and revenue share
total_rev = rfm_raw['monetary'].sum()
summary = rfm_raw.groupby('segment').agg(
    customer_count=('customer_id', 'count'),
    total_revenue=('monetary', 'sum'),
    avg_recency=('recency', 'mean'),
    avg_frequency=('frequency', 'mean'),
    avg_monetary=('monetary', 'mean')
).reset_index()

summary['customer_share_pct'] = (summary['customer_count'] / len(rfm_raw)) * 100
summary['revenue_share_pct'] = (summary['total_revenue'] / total_rev) * 100

print("=== RFM Customer Segmentation Summary ===")
print(summary.to_string(index=False))

# 6. Export all required tables to clean Excel assets
con.sql("SELECT * FROM product_metrics").df().to_excel("product_metrics.xlsx", index=False)
con.sql("SELECT * FROM category_metrics").df().to_excel("category_metrics.xlsx", index=False)
con.sql("SELECT * FROM customer_metrics").df().to_excel("customer_metrics.xlsx", index=False)
rfm_raw.to_excel("rfm_segments.xlsx", index=False)

print("\\nAll analytical Excel sheets generated successfully:")
print("- product_metrics.xlsx")
print("- category_metrics.xlsx")
print("- customer_metrics.xlsx")
print("- rfm_segments.xlsx")
```
